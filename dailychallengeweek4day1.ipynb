Cellule de Texte n°3 : Les 3 types de ML

Les trois grands types de Machine Learning
L'Apprentissage Supervisé
Définition simple :On apprend à la machine en lui donnant les questions ET les bonnes réponses données étiquetées. Elle doit comprendre la logique pour appliquer la bonne réponse à de nouvelles questions.
Mon exemple : Prédire le prix d'une maison. On donne au modèle la taille de la maison entrée et son vrai prix de vente étiquette/réponse.

L'Apprentissage Non Supervisé
Définition simple : On donne à la machine des données en vrac sans lui dire ce qu'elles contiennent données non étiquetées. C'est à elle de trouver des points communs et de regrouper les données qui se ressemblent.
Mon exemple  La segmentation de clients. Le modèle analyse les données d'un magasin et crée tout seul des groupes de clients exemple les acheteurs du dimanche, les gros dépensiers sans qu'on lui ait demandé avant.


Cellule de Texte n°4 : Le Processus de développement

Le processus de création d'un modèle de ML
Pour créer un modèle, j'ai appris qu'il faut suivre un chemin logique en 3 étapes principales :

1-La Sélection des Caractéristiques Feature Selection C'est le tri de départ. On choisit uniquement les colonnes de nos données qui sont vraiment utiles pour répondre à notre problème. On jette le reste pour faire gagner du temps à l'ordinateur et éviter les erreurs.
2-La Sélection du Modèle  On choisit le bon algorithme mathématique selon notre problème. Si on veut deviner un prix un chiffre continu, on choisit une régression. Si on veut classer en Oui ou Non, on choisit une classification.
3-L'Évaluation du Modèle On teste notre modèle sur des données qu'il n'a jamais vues pour vérifier s'il est performant. On utilise des notes comme le score de précision pour savoir si notre modèle se trompe souvent ou pas.


Schéma visuel du processus de développement
Voici la représentation graphique des trois étapes indispensables expliquées

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 3))
ax.axis('off')

box_style = dict(boxstyle="round,pad=0.5", fc="lightblue", ec="navy", lw=2)

ax.text(0.15, 0.5, "1. Sélection des\nCaractéristiques\n(Choix des variables)", ha="center", va="center", bbox=box_style, fontsize=10)
ax.text(0.50, 0.5, "2. Sélection du\nModèle\n(Choix de l'algorithme)", ha="center", va="center", bbox=box_style, fontsize=10)
ax.text(0.85, 0.5, "3. Évaluation du\nModèle\n(Calcul des scores)", ha="center", va="center", bbox=box_style, fontsize=10)

ax.annotate('', xy=(0.36, 0.5), xytext=(0.29, 0.5), arrowprops=dict(arrowstyle="->", color="navy", lw=2))
ax.annotate('', xy=(0.71, 0.5), xytext=(0.64, 0.5), arrowprops=dict(arrowstyle="->", color="navy", lw=2))

plt.title("L'Organigramme du Processus d'un Modèle ML ", fontsize=14, fontweight='bold', color="navy")
plt.show()
```
