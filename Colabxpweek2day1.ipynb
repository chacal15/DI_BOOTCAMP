Qu'est-ce que l'analyse de données ?


L’analyse de données est le processus qui consiste à collecter, organiser, examiner et interpréter des informations afin d’en tirer des conclusions utiles. Elle permet de comprendre des tendances, résoudre des problèmes et aider à la prise de décision dans plusieurs domaines comme la santé, le commerce.

Pourquoi l'analyse des données est-elle importante dans les contextes modernes ?

L’analyse des données est importante dans les contextes modernes car elle aide les entreprises et les organisations à prendre de meilleures décisions grâce aux informations recueillies. Elle permet d’identifier des tendances, d’améliorer les performances, de prévoir certains problèmes et de mieux comprendre les besoins des utilisateurs dans des domaines variés comme la santé, le commerce.

Exercice 5 : Compétences de base en observation et analyse de données

```
import matplotlib.pyplot as plt
import seaborn as sns

# Group by Year and calculate the mean of 'Avg hrs per day sleeping'
daily_sleep_by_year = df_sleeping.groupby('Year')['Avg hrs per day sleeping'].mean().reset_index()

# Create the line plot
plt.figure(figsize=(10, 6))
sns.lineplot(x='Year', y='Avg hrs per day sleeping', data=daily_sleep_by_year, marker='o')
plt.title('Évolution du sommeil moyen par année aux États-Unis')
plt.xlabel('Année')
plt.ylabel('Moyenne d\'heures de sommeil par jour')
plt.grid(True)
plt.xticks(daily_sleep_by_year['Year'])
plt.tight_layout()
plt.show()
```

Exercice 7 : Exercice de transformation


# Conversion de données non structurées en données structurées

## 1. Articles de blog sur les expériences de voyage
**Méthode :** Traitement du langage naturel (NLP) + extraction d’informations

**Justification :**
Les articles sont du texte libre. Le NLP permet d’extraire des informations importantes comme :
- destination
- dates
- sentiments (positif/négatif)
- lieux visités  
Ces données peuvent ensuite être organisées dans un tableau structuré.

---

## 2. Enregistrements audio des appels au service client
**Méthode :** Reconnaissance vocale (Speech-to-Text) + classification

**Justification :**
L’audio est d’abord transformé en texte, puis analysé pour extraire :
- motif de l’appel
- durée
- niveau de satisfaction
Cela permet de créer une base de données structurée pour l’analyse client.

---

## 3. Notes manuscrites issues d’une séance de brainstorming
**Méthode :** OCR (reconnaissance optique de caractères)

**Justification :**
L’OCR convertit l’écriture manuscrite en texte numérique.
Ensuite, on peut organiser les idées en :
- thèmes
- catégories
- priorités  
Cela facilite le classement et l’exploitation des idées.

---

## 4. Tutoriel vidéo sur la cuisine
**Méthode :** Analyse vidéo + transcription audio + extraction de contenu

**Justification :**
La vidéo est convertie en données exploitables via :
- transcription audio → texte
- extraction des étapes de la recette
- identification des ingrédients  
On obtient ainsi une recette structurée (ingrédients + étapes).

Exercice 9 : Exporter un dataframe au format Excel et au format JSON.


```
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Los Angeles', 'Chicago']
}
df_export = pd.DataFrame(data)

print("DataFrame created:")
display(df_export)
```

### Exporting the DataFrame to Excel

```
excel_file_path = '/content/output.xlsx'
df_export.to_excel(excel_file_path, index=False)

print(f"DataFrame exported to {excel_file_path}")
```

### Exporting the DataFrame to JSON

```

json_file_path = '/content/output.json'
df_export.to_json(json_file_path, orient='records', indent=4)

print(f"DataFrame exported to {json_file_path}")
```
