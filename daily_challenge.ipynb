### 2. Tokenisation du texte

**Objectif** : Comprendre comment tokeniser un texte à l'aide de tokeniseurs pré-entraînés.

**Instructions** :

* Utilisez les fonctions `BertTokenizer` et `XLMRobertaTokenizer` pour convertir les phrases en entrées tokenisées.
* Explorez les différents types de jetons, tels que `input_ids`, `attention_mask`, et `labels`.
* Expérimentez la tokenisation par phrase unique et par deux phrases.

**Fonctions à utiliser** :

* `tokenizer.encode_plus()`
* `tokenizer.decode()`

```python
# Initialisation d'un tokenizer BERT
bert_tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# Exemple de phrase unique
sentence_1 = "Ceci est un exemple de phrase pour la tokenisation."
encoded_bert = bert_tokenizer(
    sentence_1,
    add_special_tokens=True, # Ajouter des jetons spéciaux (CLS, SEP)
    max_length=64,           # Tronquer/compléter à la longueur maximale
    padding='max_length',  # Compléter à la longueur maximale
    truncation=True,         # Assurer la troncature si la phrase est trop longue
    return_attention_mask=True, # Retourner le masque d'attention
    return_tensors='pt'      # Retourner des tensors PyTorch
)

print("--- Tokenisation BERT (phrase unique) ---")
print(f"Phrase originale: {sentence_1}")
print(f"Input IDs: {encoded_bert['input_ids'][0]}")
print(f"Attention Mask: {encoded_bert['attention_mask'][0]}")
print(f"Tokens décodés: {bert_tokenizer.decode(encoded_bert['input_ids'][0])}")

# Exemple de deux phrases
sentence_2_a = "Google Colab est un outil formidable."
sentence_2_b = "Il facilite le développement de l'IA."

encoded_pair_bert = bert_tokenizer(
    sentence_2_a,
    sentence_2_b,
    add_special_tokens=True,
    max_length=64,
    padding='max_length',
    truncation=True,
    return_attention_mask=True,
    return_tensors='pt'
)

print("\n--- Tokenisation BERT (deux phrases) ---")
print(f"Phrase 1: {sentence_2_a}")
print(f"Phrase 2: {sentence_2_b}")
print(f"Input IDs: {encoded_pair_bert['input_ids'][0]}")
print(f"Attention Mask: {encoded_pair_bert['attention_mask'][0]}")
print(f"Tokens décodés: {bert_tokenizer.decode(encoded_pair_bert['input_ids'][0])}")

# Initialisation d'un tokenizer XLM-RoBERTa
xmroberta_tokenizer = XLMRobertaTokenizer.from_pretrained('xlm-roberta-base')

# Exemple de phrase unique avec XLM-RoBERTa
sentence_3 = "Ceci est une phrase en français."
encoded_xlmroberta = xmroberta_tokenizer(
    sentence_3,
    add_special_tokens=True,
    max_length=64,
    padding='max_length',
    truncation=True,
    return_attention_mask=True,
    return_tensors='pt'
)

print("\n--- Tokenisation XLM-RoBERTa (phrase unique) ---")
print(f"Phrase originale: {sentence_3}")
print(f"Input IDs: {encoded_xlmroberta['input_ids'][0]}")
print(f"Attention Mask: {encoded_xlmroberta['attention_mask'][0]}")
print(f"Tokens décodés: {xmroberta_tokenizer.decode(encoded_xlmroberta['input_ids'][0])}")
```

### 3. Préparation des données d'entrée pour le modèle

**Objectif** : Formater correctement les données d’entrée pour les modèles de transformateurs.

**Instructions** :

* Veillez à ce que les phrases d'entrée soient complétées et éventuellement tronquées à la longueur maximale.
* Comprendre et définir des jetons spéciaux tels que `<s>` et `</s>`.
* Découvrez `attention_mask` comment cela permet au modèle d'ignorer les jetons de remplissage.

**Fonctions à utiliser** :

* `tokenizer.encode_plus()` (or simply calling the tokenizer directly)
* `tokenizer.special_tokens_map`
* `tokenizer.vocab_size`

```python
# Réutiliser les tokenizers et les phrases des étapes précédentes
# bert_tokenizer et xmroberta_tokenizer sont déjà initialisés et disponibles.

# Exploration des jetons spéciaux pour BERT
print("--- Jetons spéciaux BERT ---")
print(bert_tokenizer.special_tokens_map)
print(f"Taille du vocabulaire BERT: {bert_tokenizer.vocab_size}")

# Exploration des jetons spéciaux pour XLM-RoBERTa
print("\n--- Jetons spéciaux XLM-RoBERTa ---")
print(xmroberta_tokenizer.special_tokens_map)
print(f"Taille du vocabulaire XLM-RoBERTa: {xmroberta_tokenizer.vocab_size}")

# Démonstration de l'impact de padding, truncation et attention_mask
sentence_long = "Ceci est une phrase très longue qui dépasse la longueur maximale de 15 jetons pour illustrer la troncature et le padding. Normalement, la longueur maximale est bien plus grande pour les modèles réels." # A very long sentence
max_len_example = 15 # Définir une petite longueur maximale pour l'exemple

print(f"\n--- Préparation des données avec max_length={max_len_example} ---")

# Tokenisation BERT avec troncature et padding
encoded_bert_prep = bert_tokenizer(
    sentence_long,
    max_length=max_len_example,
    padding='max_length',
    truncation=True,
    return_attention_mask=True,
    return_tensors='pt'
)

print("\n--- BERT --- ")
print(f"Phrase originale (longue): {sentence_long}")
print(f"Input IDs (BERT): {encoded_bert_prep['input_ids'][0]}")
print(f"Attention Mask (BERT): {encoded_bert_prep['attention_mask'][0]}")
print(f"Tokens décodés (BERT): {bert_tokenizer.decode(encoded_bert_prep['input_ids'][0])}")

# Tokenisation XLM-RoBERTa avec troncature et padding
encoded_xlmroberta_prep = xmroberta_tokenizer(
    sentence_long,
    max_length=max_len_example,
    padding='max_length',
    truncation=True,
    return_attention_mask=True,
    return_tensors='pt'
)

print("\n--- XLM-RoBERTa --- ")
print(f"Phrase originale (longue): {sentence_long}")
print(f"Input IDs (XLM-RoBERTa): {encoded_xlmroberta_prep['input_ids'][0]}")
print(f"Attention Mask (XLM-RoBERTa): {encoded_xlmroberta_prep['attention_mask'][0]}")
print(f"Tokens décodés (XLM-RoBERTa): {xmroberta_tokenizer.decode(encoded_xlmroberta_prep['input_ids'][0])}")
```

### 5. Création de plis de validation croisée

**Objectif** : Mettre en œuvre une validation croisée à k plis pour l'entraînement.

**Instructions** :

* À utiliser `StratifiedKFold` pour créer 5 ensembles d'entraînement et de validation.
* Veillez à ce que chaque pli conserve la même répartition des étiquettes.
* Stockez les ensembles d'entraînement et de validation dans des listes séparées.

**Fonctions à utiliser** :

* `from sklearn.model_selection import StratifiedKFold`
* `kf.split()`
* `StratifiedKFold(shuffle=True)`

```python
from sklearn.model_selection import StratifiedKFold

# Définir le nombre de plis (folds)
n_splits = 5

# Initialiser StratifiedKFold
kf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)

# Listes pour stocker les indices des plis
train_indices_folds = []
val_indices_folds = []

# Appliquer la validation croisée stratifiée
# Assurez-vous d'utiliser la colonne 'label' pour la stratification
for fold, (train_index, val_index) in enumerate(kf.split(df_train, df_train['label'])):
    train_indices_folds.append(train_index)
    val_indices_folds.append(val_index)
    print(f"Plis {fold + 1}/{n_splits} - Taille d'entraînement: {len(train_index)}, Taille de validation: {len(val_index)}")

    # Vérifier la distribution des labels dans chaque pli (optionnel)
    train_labels = df_train.iloc[train_index]['label']
    val_labels = df_train.iloc[val_index]['label']
    print(f"  Distribution des labels d'entraînement:\n{train_labels.value_counts(normalize=True)}")
    print(f"  Distribution des labels de validation:\n{val_labels.value_counts(normalize=True)}")

print(f"\nValidation croisée en {n_splits} plis créée avec succès.")
```
