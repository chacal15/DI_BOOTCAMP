Exemples concrets

ML traditionnel plus performant :La prédiction du prix d'un bien immobilier à partir de quelques variables structurées (surface, nombre de pièces, localisation, année de construction) se prête bien à une régression linéaire ou à une forêt aléatoire. Apprentissage profond plus performant :La reconnaissance d'objets dans des images de rues pour un véhicule autonome nécessite l'apprentissage profond, en particulier des réseaux de neurones convolutifs. Paragraphe : avantage de l'apprentissage profond pour les données non structurées

L'apprentissage profond présente un avantage majeur dans le traitement des données non structurées comme les images, le texte ou l'audio, car il repose sur l'apprentissage de représentations. Plutôt que de dépendre de caractéristiques définies manuellement, les réseaux de neurones profonds construisent eux-mêmes une hiérarchie de représentations internes, des plus simples aux plus abstraites. Dans un réseau convolutif par exemple, les premières couches détectent des motifs élémentaires comme des contours ou des textures, tandis que les couches plus profondes combinent ces éléments pour reconnaître des formes, puis des objets entiers : ce sont les **caractéristiques hiérarchiques. Cette capacité à apprendre automatiquement des niveaux d'abstraction croissants permet au modèle de capturer des relations complexes et non linéaires que l'ingénierie manuelle de caractéristiques ne pourrait jamais anticiper. En conséquence, l'apprentissage profond s'adapte naturellement à la richesse et à la variabilité des données non structurées, là où le ML traditionnel atteint rapidement ses limites.

## 🌟 Exercise 5: Cross-Validation to Find the Optimal Degree

**As stated in the exercise**  
Objective: Use cross-validation to determine the best polynomial degree.  

Instructions: Loop degrees from 1 to 11. For each degree, compute the RMSE on the training and test sets. Store results in a list. Plot RMSE vs degree with a logarithmic y axis. Identify the degree minimizing test RMSE and confirm it matches the true model \(y=-x^2\).


```python
# To-Do: compute RMSE per degree for train and test
# def rmse(y_true, y_pred): ...
# degrees = range(1, 12)
# rows = []  # each row: (degree, rmse_train, rmse_test)
# for d in degrees:
#     # fit on train
#     # evaluate on train and test
#     # append to rows
#     pass
# rows[:3]  # peek
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# 1. Définition de la fonction RMSE
def rmse(y_true, y_pred):
    return np.sqrt(mean_squared_error(y_true, y_pred))

# Configuration des degrés (de 1 à 11 inclus)
degrees = range(1, 12)
rows = []  # Chaque ligne stockera : (degré, rmse_train, rmse_test)

# 2. Boucle de validation croisée / évaluation
for d in degrees:
    # Transformation polynomiale des données
    poly = PolynomialFeatures(degree=d)
    X_train_poly = poly.fit_transform(x_train.reshape(-1, 1))
    X_test_poly = poly.transform(x_test.reshape(-1, 1))

    # Entraînement du modèle (Fit on train)
    model = LinearRegression()
    model.fit(X_train_poly, y_train)

    # Prédictions
    y_train_pred = model.predict(X_train_poly)
    y_test_pred = model.predict(X_test_poly)

    # Évaluation (Evaluate on train and test)
    rmse_train = rmse(y_train, y_train_pred)
    rmse_test = rmse(y_test, y_test_pred)

    # Stockage des résultats
    rows.append((d, rmse_train, rmse_test))

# Aperçu des 3 premières lignes
print("Peek rows[:3]:", rows[:3])

# =====================================================================
# Code de visualisation (Fourni dans votre exercice)
# =====================================================================
degs = [r[0] for r in rows]
rmse_tr = [r[1] for r in rows]
rmse_te = [r[2] for r in rows]

plt.figure(figsize=(6, 4))
plt.plot(degs, rmse_tr, label="train RMSE", marker='o')
plt.plot(degs, rmse_te, label="test RMSE", marker='o')
plt.yscale("log")
plt.xlabel("polynomial degree")
plt.ylabel("RMSE")
plt.title("RMSE vs degree")
plt.legend()
plt.tight_layout()
plt.show()
```
