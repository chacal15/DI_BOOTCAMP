### Exercise 2: Artificial Neural Networks (ANNs) - Diagram Description

**To-Do:** Below is a textual description of the ANN diagram as requested in the exercise. The diagram consists of:

*   **Input Layer:** 3 neurons (representing features like $x_1, x_2, x_3$)
*   **Hidden Layer:** 4 neurons (processing inputs)
*   **Output Layer:** 2 neurons (producing final predictions)

**Components Labeled:**
*   **Neurons:** Represented by circles in each layer.
*   **Weights ($w$):** Lines connecting neurons between layers, each with an associated numerical value.
*   **Biases ($b$):** An additional parameter for each neuron in the hidden and output layers, shifting the activation function's output.
*   **Activation Function ($f$):** A non-linear function applied to the weighted sum plus bias within each hidden and output neuron.
*   **Layers:** Input, Hidden, and Output layers.

For the ANN diagram, since I cannot directly draw, I will describe the components and their connections as per your instructions:

**Artificial Neural Network Diagram Description:**

**Input Layer (3 Neurons):** These neurons receive the initial data. Let's call their values $x_1, x_2, x_3$. They simply pass the input values forward.

**Hidden Layer (4 Neurons):** Each neuron in this layer receives input from all neurons in the input layer. For any hidden neuron $h_j$, its input is a **weighted sum** of the input layer's activations plus a **bias** term. Mathematically, $z_j = (w_{1j}x_1 + w_{2j}x_2 + w_{3j}x_3) + b_j$, where $w_{ij}$ are the **weights** connecting input neuron $i$ to hidden neuron $j$, and $b_j$ is the **bias** for hidden neuron $j$. This sum then passes through an **activation function** (e.g., ReLU, sigmoid) to produce the neuron's output. The hidden layer's output is then passed to the output layer.

**Output Layer (2 Neurons):** Similar to the hidden layer, each output neuron $o_k$ receives a weighted sum of the hidden layer's activations plus its own bias term. This sum is then passed through an **activation function** (e.g., softmax for classification, linear for regression) to produce the final **outputs** of the network, $y_1, y_2$.

### Role of Each Layer in the ANN

Let's break down the role of each layer in the described Artificial Neural Network:

1.  **Input Layer:**
    *   **Role:** The input layer is the entry point of the network. Its neurons receive the raw input data. In our example, it has 3 neurons, meaning it takes 3 distinct features or values as input (e.g., $x_1, x_2, x_3$).
    *   **Function:** These neurons typically do not perform any complex computations (like applying activation functions) themselves. Their primary job is to pass the input values directly to the next layer (the hidden layer).

2.  **Hidden Layer:**
    *   **Role:** The hidden layer(s) are where the majority of the computational processing and feature extraction occurs. They are 'hidden' because they are not directly exposed to the external input or output.
    *   **Function:** Each neuron in this layer:
        *   **Receives Weighted Inputs:** Takes inputs from all neurons in the previous layer (the input layer in this case), multiplying each input by an associated **weight** ($w$).
        *   **Sums Inputs with Bias:** Adds all these weighted inputs together and then adds a **bias** term ($b$). This creates a "weighted sum" (or net input).
        *   **Applies Activation Function:** Passes this weighted sum through a non-linear **activation function** ($f$, e.g., ReLU, sigmoid). This non-linearity is crucial for the network to learn complex patterns and relationships in the data that go beyond simple linear combinations.
        *   **Outputs to Next Layer:** The output of the activation function for each hidden neuron then serves as input to the next layer (the output layer).

3.  **Output Layer:**
    *   **Role:** The output layer is the final layer of the network, responsible for producing the network's predictions or decisions. In our example, it has 2 neurons, meaning it produces 2 output values (e.g., $y_1, y_2$).
    *   **Function:** Similar to the hidden layer, each output neuron performs a weighted sum of its inputs from the hidden layer, adds a bias, and then applies an activation function. The choice of activation function here depends on the type of task:
        *   For **regression** (predicting continuous values), a linear activation function (or no activation) is often used.
        *   For **classification** (predicting categories), activation functions like sigmoid (for binary classification) or softmax (for multi-class classification) are common to produce probabilities.

**Learning point**  
An ANN composes affine transformations with nonlinear activations. The affine part mixes features through weights and biases. The nonlinearity enables modeling of complex decision boundaries.

## 🌟 Exercise 4: Fitting Polynomial Models of Different Degrees

**As stated in the exercise**  
Objective: Fit polynomial models of varying degrees and observe overfitting.  

Instructions: Define `polynomial_fit(degree)` that returns the fitted polynomial coefficients from training data. Define `plot_polyfit(degree)` that draws the training set, the test set, and the fitted polynomial curve. Visualize fits for degrees 1, 7, and 11. Observe behavior on training and test sets as degree increases.


```python
def polynomial_fit(degree):
    coeffs = np.polyfit(x_train, y_train, degree)
    return coeffs
```

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_polyfit(degree):
    coeffs = polynomial_fit(degree)

    x_line = np.linspace(min(x), max(x), 200)
    y_line = np.polyval(coeffs, x_line)

    plt.figure(figsize=(6,4))
    plt.scatter(x_train, y_train, label="Train")
    plt.scatter(x_test, y_test, label="Test")
    plt.plot(x_line, y_line)

    plt.title(f"Polynomial degree {degree}")
    plt.legend()
    plt.show()
```

```python
x_train, y_train = x[:12], y[:12]
x_test,  y_test  = x[-8:], y[-8:]

plot_polyfit(1)
plot_polyfit(7)
plot_polyfit(11)
```

**Learning point**

## 🌟 Exercise 5: Cross-Validation to Find the Optimal Degree

**As stated in the exercise**  
Objective: Use cross-validation to determine the best polynomial degree.  

Instructions: Loop degrees from 1 to 11. For each degree, compute the RMSE on the training and test sets. Store results in a list. Plot RMSE vs degree with a logarithmic y axis. Identify the degree minimizing test RMSE and confirm it matches the true model \(y=-x^2\).


```python
# To-Do: compute RMSE per degree for train and test
# def rmse(y_true, y_pred): ...
# degrees = range(1, 12)
# rows = []  # each row: (degree, rmse_train, rmse_test)
# for d in degrees:
#     # fit on train
#     # evaluate on train and test
#     # append to rows
#     pass
# rows[:3]  # peek
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# 1. Définition de la fonction RMSE
def rmse(y_true, y_pred):
    return np.sqrt(mean_squared_error(y_true, y_pred))

# Configuration des degrés (de 1 à 11 inclus)
degrees = range(1, 12)
rows = []  # Chaque ligne stockera : (degré, rmse_train, rmse_test)

# 2. Boucle de validation croisée / évaluation
for d in degrees:
    # Transformation polynomiale des données
    poly = PolynomialFeatures(degree=d)
    X_train_poly = poly.fit_transform(x_train.reshape(-1, 1))
    X_test_poly = poly.transform(x_test.reshape(-1, 1))

    # Entraînement du modèle (Fit on train)
    model = LinearRegression()
    model.fit(X_train_poly, y_train)

    # Prédictions
    y_train_pred = model.predict(X_train_poly)
    y_test_pred = model.predict(X_test_poly)

    # Évaluation (Evaluate on train and test)
    rmse_train = rmse(y_train, y_train_pred)
    rmse_test = rmse(y_test, y_test_pred)

    # Stockage des résultats
    rows.append((d, rmse_train, rmse_test))

# Aperçu des 3 premières lignes
print("Peek rows[:3]:", rows[:3])

# =====================================================================
# Code de visualisation (Fourni dans votre exercice)
# =====================================================================
degs = [r[0] for r in rows]
rmse_tr = [r[1] for r in rows]
rmse_te = [r[2] for r in rows]

plt.figure(figsize=(6, 4))
plt.plot(degs, rmse_tr, label="train RMSE", marker='o')
plt.plot(degs, rmse_te, label="test RMSE", marker='o')
plt.yscale("log")
plt.xlabel("polynomial degree")
plt.ylabel("RMSE")
plt.title("RMSE vs degree")
plt.legend()
plt.tight_layout()
plt.show()
```

**To-Do:** Report the degree with the minimum test RMSE and explain in 2–3 sentences how this connects to the true model \(y=-x^2\) and the bias–variance trade off.

From the generated plot and the `rows` data, we can observe the following:

Looking at the test RMSE values, the minimum test RMSE occurs at **degree 2**. The true model generating the data was \(y = -x^2\), which is a polynomial of degree 2. This perfectly aligns with our cross-validation result, indicating that the model successfully identified the underlying complexity of the data without overfitting or underfitting.

This demonstrates the bias-variance trade-off: a degree-1 model (linear) would have high bias (underfitting) as it cannot capture the quadratic relationship. Higher-degree models (e.g., degree 7 or 11) might achieve very low training RMSE but show much higher test RMSE (overfitting), as they learn the noise in the training data rather than the true underlying pattern. Degree 2 provides the optimal balance, having low bias (capturing the true function) and low variance (not overly sensitive to the training data's noise).
