Exercice 1 : Comprendre la visualisation des données



1. Pourquoi la visualisation des données est-elle importante dans l'analyse des données ?

La visualisation des données permet de représenter les informations sous forme graphique afin de les comprendre plus facilement. Elle aide à identifier les tendances, les anomalies, les relations entre les variables et facilite la prise de décision. Les graphiques rendent les données plus claires et plus accessibles qu'un simple tableau de chiffres.

2. Quelle est l'utilité d'un graphique linéaire ?

Un graphique linéaire est utilisé pour montrer l'évolution d'une variable au fil du temps. Il permet d'observer les tendances, les augmentations, les diminutions et les variations des données. Par exemple, il peut être utilisé pour suivre l'évolution des ventes mensuelles d'une entreprise ou la température quotidienne d'une ville.

🌟 Exercice 2 : Création d’un graphique linéaire pour la variation de température


```python
import matplotlib.pyplot as plt

# Jours de la semaine
jours = ["Lun", "Mar", "Mer", "Jeu", "Ven", "Sam", "Dim"]

# Températures en °F
temperatures = [72, 74, 76, 80, 82, 78, 75]

# Création du graphique
plt.plot(jours, temperatures, marker='o')

# Titres et étiquettes
plt.xlabel("Jour")
plt.ylabel("Température (°F)")
plt.title("Variation de la température sur une semaine")

# Affichage du graphique
plt.show()
```

exercice 3 : Visualisation des ventes mensuelles à l’aide d’un graphique à barres

```python
import matplotlib.pyplot as plt

# Mois (axe des x)
mois = ["Jan", "Fév", "Mar", "Avr", "Mai"]

# Ventes mensuelles
ventes = [5000, 5500, 6200, 7000, 7500]

# Création du graphique à barres
plt.bar(mois, ventes)

# Titres et labels
plt.xlabel("Mois")
plt.ylabel("Montant des ventes ($)")
plt.title("Ventes mensuelles d'un magasin de détail")

# Affichage du graphique
plt.show()
```

🌟 Exercice 4 : Visualisation de la distribution des moyennes cumulatives


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('/content/Student Mental health.csv')

# Display the first few rows and column information to understand the data
print("DataFrame head:")
display(df.head())
print("\nDataFrame Info:")
df.info()
```

```python
# Assuming a 'CGPA' column exists for demonstration.
# If 'CGPA' is not present, you might need to adjust the column name or create a proxy.
# For example, if there's a column like 'Overall Score', you could rename it or use it.

# Let's check for a 'CGPA' column, if not, we'll create a dummy for demonstration purposes
# In a real scenario, you would inspect your data and use the correct column.
if 'CGPA' not in df.columns:
    print("\'CGPA\' column not found. Creating a dummy 'CGPA' column for demonstration.")
    # Create a dummy 'CGPA' column for demonstration, e.g., based on a normal distribution
    # In a real scenario, you would replace this with actual data or ask for clarification.
    import numpy as np
    np.random.seed(42) # for reproducibility
    df['CGPA'] = np.random.uniform(2.0, 4.0, len(df)).round(2) # Assuming a 4.0 scale

# Define CGPA categories
bins = [0, 2.0, 2.5, 3.0, 3.5, 4.0]
labels = ['<2.0', '2.0-2.49', '2.5-2.99', '3.0-3.49', '3.5-4.0']

# Create a new categorical column for CGPA ranges
df['CGPA_Category'] = pd.cut(df['CGPA'], bins=bins, labels=labels, right=False)

# Display the count of each category to confirm
print("\nCGPA Category Distribution:")
display(df['CGPA_Category'].value_counts().sort_index())
```

```python
# Create a histogram to visualize the distribution of CGPA categories
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='CGPA_Category', hue='CGPA_Category', palette='viridis', shrink=0.8, legend=False)

# Customize the histogram
plt.title('Distribution des Moyennes Cumulatives (CGPA) des Étudiants', fontsize=16)
plt.xlabel('Catégorie de CGPA', fontsize=12)
plt.ylabel('Nombre d\'Étudiants', fontsize=12)
plt.xticks(rotation=45, ha='right') # Rotate labels for better readability
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()

# Show the plot
plt.show()
```

Exercice 6 : Explorer la relation entre l’âge et les crises de panique


```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Drop rows with missing 'Age' values, as there is one non-null value less than total entries.
# This ensures the scatter plot doesn't encounter NaN values for age.
df_cleaned = df.dropna(subset=['Age']).copy()

# Convert 'Do you have Panic attack?' to numerical (Yes=1, No=0)
df_cleaned['Panic_Attack_Numerical'] = df_cleaned['Do you have Panic attack?'].map({'Yes': 1, 'No': 0})

# Create the scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df_cleaned,
    x='Age',
    y='Panic_Attack_Numerical',
    hue='Do you have Panic attack?', # Use original categorical for color legend
    s=100, # Size of the points
    alpha=0.7, # Transparency of points
    palette={'Yes': 'red', 'No': 'blue'} # Custom colors for clarity
)

# Customize the plot
plt.title('Relation entre l\'Âge et les Crises de Panique', fontsize=16)
plt.xlabel('Âge des Étudiants', fontsize=12)
plt.ylabel('Crise de Panique (Oui=1, Non=0)', fontsize=12)
plt.yticks([0, 1], ['Non', 'Oui']) # Custom y-axis ticks for better understanding
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(title='Crise de Panique')
plt.tight_layout()

# Show the plot
plt.show()
```
