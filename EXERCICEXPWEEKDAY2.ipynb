Exercice 1 : Opérations matricielles

```
import numpy as np

matrice = np.array([
    [2, 1, 1],
    [1, 3, 2],
    [1, 0, 0]
])

print("Matrice de base :")
print(matrice)

det = np.linalg.det(matrice)
print("\nDeterminant :", det)

if det != 0:
    inverse = np.linalg.inv(matrice)
    print("\nInverse de la matrice :")
    print(inverse)
else:
    print("\nLa matrice n'est pas inversible.")
```

Exercice 2 : Analyse statistique

```
import numpy as np

np.random.seed(42)
donnees = np.random.randint(1, 101, size=50)

moyenne = np.mean(donnees)

mediane = np.median(donnees)

ecart_type = np.std(donnees)

print(f"Tableau de données :\n{donnees}\n")
print(f"La moyenne est : {moyenne:.2f}")
print(f"La médiane est : {mediane:.1f}")
print(f"L'écart type est : {ecart_type:.2f}")
```

Exercice 4 : Manipulation de données avec NumPy et Pandas


```
import pandas as pd
import numpy as np

data = np.random.randint(1, 100, size=(10, 3))
df = pd.DataFrame(data, columns=['X', 'Y', 'Z'])
print("Donnees de depart :")
print(df)

filtre = df[df['X'] > 50]
print("\nLignes ou X > 50 :")
print(filtre)

somme_total = df.sum()
moyenne_total = df.mean()

print("\nSomme par colonne :")
print(somme_total)

print("\nMoyenne par colonne :")
print(moyenne_total)
```

```
import numpy as np

productivity_before = np.random.normal(loc=50, scale=10, size=30)
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

moy_avant = np.mean(productivity_before)
moy_apres = np.mean(productivity_after)
evolution = moy_apres - moy_avant

print("Moyenne avant :", moy_avant)
print("Moyenne apres :", moy_apres)
print("Evolution globale :", evolution)

if evolution > 0:
    print("Resultat : La formation ameliore la productivite.")
else:
    print("Resultat : Aucun impact significatif de la formation.")
```

Exercice 8 : Manipulation de données de séries temporelles

```
import pandas as pd
import numpy as np

dates = pd.date_range(start='2023-01-01', end='2023-12-31', freq='D')
valeurs = np.random.randint(10, 100, size=len(dates))

serie = pd.Series(valeurs, index=dates)

t1 = serie['2023-01':'2023-03']
t2 = serie['2023-04':'2023-06']
t3 = serie['2023-07':'2023-09']
t4 = serie['2023-10':'2023-12']

print("Nombre de jours Janvier a Mars :", len(t1))
print("Nombre de jours Avril a Juin :", len(t2))
print("Nombre de jours Juillet a Septembre :", len(t3))
print("Nombre de jours Octobre a Decembre :", len(t4))

print("\nExtrait du premier intervalle (Janvier-Mars) :")
print(t1.head())
```

Exercice 9 : Conversion de données


```
import numpy as np
import pandas as pd

# Création d'un tableau NumPy
tableau = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

# Conversion en DataFrame Pandas
df = pd.DataFrame(tableau, columns=['A', 'B', 'C'])

print("DataFrame obtenu :")
print(df)
```

DataFrame permet de transformer un tableau NumPy en DataFrame Pandas.
numpy permet de convertir un DataFrame Pandas en tableau NumPy.
Ces conversions sont utiles pour passer facilement entre les outils de calcul numérique NumPy et d'analyse de données Pandas.
