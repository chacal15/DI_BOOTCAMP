```python
# =========================================
# 1. IMPORTATION DES BIBLIOTHÈQUES
# =========================================
import pandas as pd
import numpy as np

from sklearn.preprocessing import MinMaxScaler
from sklearn.decomposition import PCA

# =========================================
# 2. CHARGEMENT DU DATASET
# =========================================
df = pd.read_csv("/content/datascience_salaries.csv")

# Suppression de la colonne inutile
df = df.drop(columns=["Unnamed: 0"])

print(df.head())

# =========================================
# 3. NORMALISATION MIN-MAX DU SALAIRE
# =========================================
scaler = MinMaxScaler()
df["salary_normalized"] = scaler.fit_transform(df[["salary"]])

# =========================================
# 4. ENCODAGE DES VARIABLES CATEGORIELLES
# =========================================
df_encoded = pd.get_dummies(df, drop_first=True)

# =========================================
# 5. RÉDUCTION DE DIMENSION (PCA)
# =========================================
pca = PCA(n_components=2)
pca_result = pca.fit_transform(df_encoded)

df["PCA1"] = pca_result[:, 0]
df["PCA2"] = pca_result[:, 1]

# =========================================
# 6. AGRÉGATION PAR NIVEAU D'EXPÉRIENCE
# =========================================
result = df.groupby("experience_level")["salary"].agg(["mean", "median"])

print("\nSalaire moyen et médian par niveau d'expérience :")
print(result)

# =========================================
# 7. SAUVEGARDE DU RÉSULTAT
# =========================================
df.to_csv("resultat_salaries_analyse.csv", index=False)
```
