## Exercise 1: Rewrite and Optimize a Vague Prompt
**Scenario**: Manager asks for a LinkedIn post about FlowNest with the vague prompt `"Write something about productivity tips."`

### Tasks
- Identify 3 critical issues with the prompt.
- Rewrite it using: role prompting, clear instructions, and format/tone/length constraints.
- Add a final line to make it reusable for future social posts.

```python
# Trois problemes critiques de la consigne vague
critical_issues = [
    "Aucun public cible ni voix de marque : le modele ignore a qui il parle (fondateurs ? etudiants ?) et sur quel ton FlowNest s'exprime.",
    "Aucune mention du produit ni objectif : la consigne ne dit pas qu'il s'agit de promouvoir FlowNest, ni quelle action on attend du lecteur.",
    "Aucune contrainte de format, de longueur ou de plateforme : rien n'indique qu'il s'agit d'un post LinkedIn, ni sa structure, ni sa taille.",
]
critical_issues
```

```python
# Consigne reecrite : role + instructions + contraintes + ligne de reutilisation
rewritten_prompt = """
ROLE
Tu es content strategist senior pour FlowNest, une application de concentration destinee aux
professionnels du savoir (developpeurs, designers, chefs de projet) submerges par les notifications.
La voix de la marque est : calme, concrete, bienveillante, jamais culpabilisante, zero jargon
"hustle culture".

TACHE
Redige UN post LinkedIn qui promeut FlowNest en partant d'astuces de productivite reellement
actionnables.

INSTRUCTIONS
1. Ouvre par une accroche d'une phrase decrivant une frustration concrete du public cible.
2. Donne exactement 3 actions concretes que le lecteur peut appliquer aujourd'hui, chacune en une
   phrase, en puce.
3. Relie naturellement la 3e astuce a une fonctionnalite de FlowNest (une seule mention du produit,
   pas de discours commercial).
4. Termine par une question ouverte pour declencher les commentaires, puis 3 hashtags maximum.

CONTRAINTES
- Longueur totale : 120 mots maximum.
- Ton : amical, direct, tutoiement, phrases courtes.
- Format : accroche + 3 puces + question + hashtags. Pas de titres, pas d'emojis en debut de ligne.
- Interdits : superlatifs ("revolutionnaire", "incroyable"), promesses chiffrees non verifiables.

REUTILISATION
Utilise ce meme format (role, accroche, 3 puces actionnables, une mention produit, question,
hashtags, 120 mots max) chaque fois que nous faisons la promotion d'un outil de productivite
sur les reseaux sociaux : il suffit de remplacer le nom du produit, le public cible et la
fonctionnalite mise en avant.
"""
print(rewritten_prompt)
```

## Exercise 2: Multi-Part Prompt for Quiz Generation
**Scenario**: Create an educational prompt for 7th graders about volcanic eruptions.

### Tasks
- Ask for: 2-bullet summary + 3 MCQs (1 correct, 2 distractors).
- Require friendly tone, simple vocab (11?13 y/o), and Google Slides-ready output.
- Explain 3 ways your prompt improves on: `Make a quiz for kids about this article.`

```python
# Consigne complete en plusieurs parties
quiz_prompt = """
ROLE
Tu es enseignant de sciences en 5e (eleves de 11 a 13 ans) et concepteur de supports pedagogiques.

SOURCE
Utilise UNIQUEMENT l'article ci-dessous sur les eruptions volcaniques. N'ajoute aucune information
exterieure a ce texte.

<<<ARTICLE
[COLLER ICI LE TEXTE DE L'ARTICLE]
ARTICLE>>>

TACHES
1. RESUME : 2 puces maximum, une idee cle par puce, 20 mots maximum par puce.
2. QUIZ : exactement 3 questions a choix multiples, chacune avec 4 elements :
   - la question (une phrase, moins de 20 mots)
   - 1 bonne reponse
   - 2 distracteurs plausibles mais clairement faux d'apres l'article
   - une explication de 1 phrase de la bonne reponse
   Marque la bonne reponse avec (correct) et melange sa position d'une question a l'autre.

TON ET VOCABULAIRE
- Ton amical et encourageant, on s'adresse directement a l'eleve ("tu").
- Vocabulaire simple pour 11-13 ans : aucun terme technique sans definition immediate entre
  parentheses (ex. "magma (roche fondue)").
- Phrases de 15 mots maximum.

FORMAT DE SORTIE (compatible Google Slides)
Renvoie du texte brut structure comme suit, sans Markdown, sans tableau :

Slide 1: Ce qu'il faut retenir
- puce 1
- puce 2

Slide 2: Question 1
Question: ...
A) ...
B) ... (correct)
C) ...
Explication: ...

Slide 3: Question 2
[meme structure]

Slide 4: Question 3
[meme structure]

Une slide = un titre + son contenu, pour pouvoir copier-coller directement dans Google Slides
via Fichier > Importer des diapositives / Plan.

VERIFICATION
Si une information necessaire a une question ne figure pas dans l'article, ecris
"Information non disponible dans l'article" au lieu de l'inventer.
"""
print(quiz_prompt)
```

```python
# Trois ameliorations par rapport a "Make a quiz for kids about this article."
improvements = [
    "Ancrage sur la source : la consigne impose d'utiliser uniquement l'article et de signaler l'information manquante, ce qui empeche le modele d'inventer des faits volcanologiques (hallucinations).",
    "Specification du public et du niveau de langue : 11-13 ans, ton amical, phrases de 15 mots max, definition de chaque terme technique, ce qui evite un quiz trop abstrait ou trop enfantin ('kids' pouvait signifier 5 ans comme 15 ans).",
    "Structure de sortie exigee : nombre exact de puces, de questions, de distracteurs, marquage de la bonne reponse et gabarit slide par slide, ce qui rend le resultat directement exploitable dans Google Slides sans retravail.",
]
improvements
```

## Exercise 4: Match Prompt to Purpose
Pick a style (Exploratory, Structured, Conversational, Functional) to build a customer-support chatbot prompt.

### Tasks
- Write the full prompt in the chosen style.
- Identify 2 features that prove it matches the style.
- Justify why this style fits better than the others.

```python
# Exercice 4 : style choisi et consigne complete
chosen_style = 'Structure'

prompt_template = """
ROLE
Tu es l'assistant de support client de Nova&Co, une boutique e-commerce de vetements.

TACHE
A partir du message client fourni, produis une reponse en suivant STRICTEMENT le format ci-dessous,
sans jamais en changer l'ordre ni le nombre de sections.

MESSAGE CLIENT :
<<<{message_client}>>>
DONNEES COMMANDE : {statut_commande}

FORMAT DE REPONSE (obligatoire) :
1. Accuse de reception : 1 phrase qui reformule la demande du client.
2. Diagnostic : 1 phrase indiquant la cause identifiee, uniquement a partir des DONNEES COMMANDE.
3. Etapes de resolution : exactement 3 puces numerotees, chacune une action concrete, verbe a
   l'infinitif en tete.
4. Delai : 1 phrase avec une date ou un nombre de jours ouvres.
5. Cloture : 1 phrase proposant un canal de contact.

CONTRAINTES
- 120 mots maximum au total.
- Ton poli, neutre, vouvoiement.
- Aucune promesse de remboursement ou de geste commercial non present dans les DONNEES COMMANDE.
- Si une information manque, ecris a l'etape concernee : "Information a verifier par un conseiller."
"""

key_style_features = [
    "Gabarit fixe et numerote (5 sections imposees, exactement 3 puces d'action) : la sortie est previsible et identique d'un ticket a l'autre.",
    "Contraintes mesurables et verifiables (120 mots max, verbe a l'infinitif, delai chiffre, phrase de repli si donnee manquante) plutot qu'une intention de ton.",
]

justification = (
    "Un chatbot de support traite des milliers de tickets et doit etre auditable, integrable dans une "
    "interface et coherent avec la marque : le style Structure garantit qu'on retrouve toujours accuse de "
    "reception, diagnostic, actions et delai, ce qui reduit les oublis et facilite le controle qualite. "
    "Le style Exploratoire generait des options divergentes la ou le client attend une reponse unique. "
    "Le style Conversationnel produirait un ton agreable mais une longueur et un contenu imprevisibles, "
    "avec un risque de promesses non tenues. Le style Fonctionnel conviendrait a une simple transformation "
    "de donnees, mais pas a une reponse client qui doit aussi rassurer et orienter."
)

print(chosen_style)
print(prompt_template)
```

## Exercise 6: Hallucination Spotting and Mitigation
**Scenario**: Model added an unsupported claim in a climate/marine article summary.

### Tasks
1) Prompt the model to avoid hallucinations using only provided text.
2) Rewrite with verification language.
3) List 2 mitigation strategies.
4) Name 2 domains where hallucinations are risky and why.

```python
# Consigne anti-hallucination de base
no_hallucination_prompt = """
Tu es assistant de synthese scientifique.

Resume l'article evalue par les pairs ci-dessous, portant sur le changement climatique et la
biodiversite marine.

REGLE ABSOLUE : utilise UNIQUEMENT le texte fourni entre les balises ARTICLE. N'utilise aucune
connaissance exterieure, aucune statistique, aucune projection et aucun nom d'auteur qui ne soit
pas ecrit noir sur blanc dans ce texte.

<<<ARTICLE
[COLLER L'ARTICLE ICI]
ARTICLE>>>

Format : 5 puces maximum. Pour chaque chiffre ou affirmation, cite entre guillemets le fragment
exact de l'article (moins de 10 mots) qui l'appuie.

Si une information demandee n'apparait pas dans l'article, ecris exactement :
"Information non disponible dans l'article." Ne comble jamais un vide par une estimation.
"""
print(no_hallucination_prompt)
```

```python
# Reecriture avec langage de verification
verification_prompt = """
Tu es assistant de synthese scientifique, evalue sur ta fidelite a la source et non sur ton style.

ETAPE 1 - EXTRACTION
Liste chaque affirmation factuelle de l'article (chiffres, projections, especes, dates) avec le
fragment de phrase source correspondant.

ETAPE 2 - VERIFICATION
Pour chaque affirmation que tu envisages d'ecrire dans le resume, reponds a la question :
"Cette affirmation est-elle explicitement mentionnee dans l'article ?"
- Si OUI : conserve-la et joins le fragment source.
- Si NON ou en cas de doute : supprime-la et note "Information non disponible dans l'article."
N'inclus que les affirmations explicitement mentionnees dans l'article.

ETAPE 3 - RESUME
Redige le resume final (5 puces maximum) uniquement a partir des affirmations validees a l'etape 2.

ETAPE 4 - AUTOCONTROLE
Relis ton resume et signale sous le titre "Elements ecartes :" toute affirmation que tu as retiree
faute de support textuel. Si tu as ecrit une projection chiffree (ex. un pourcentage d'especes
disparues d'ici une date), verifie qu'elle figure mot pour mot dans l'article ; sinon supprime-la.

<<<ARTICLE
[COLLER L'ARTICLE ICI]
ARTICLE>>>
"""
print(verification_prompt)
```

```python
# Strategies d'attenuation et domaines a risque
mitigation_strategies = [
    "Ancrage strict sur la source + citation obligatoire : le modele doit joindre a chaque affirmation le fragment exact de l'article qui l'appuie, ce qui rend l'invention immediatement visible et couteuse.",
    "Clause de repli explicite et autocontrole en etapes : on autorise et on impose la reponse 'Information non disponible dans l'article', puis on demande une relecture qui liste les elements ecartes, plutot que de laisser le modele combler les vides.",
]

risky_domains = {
    'domain_1': {
        'name': 'Sante / medecine',
        'why': "Une posologie, une interaction medicamenteuse ou un diagnostic inventes peuvent conduire a un traitement inadapte et causer un prejudice corporel irreversible ; l'utilisateur, non expert, n'a souvent aucun moyen de detecter l'erreur.",
    },
    'domain_2': {
        'name': 'Juridique',
        'why': "Une jurisprudence, un article de loi ou un delai de prescription hallucines peuvent faire perdre un proces, entrainer des sanctions disciplinaires pour l'avocat et priver une personne de ses droits ; les references fabriquees paraissent credibles car elles respectent le format des citations reelles.",
    },
}

mitigation_strategies, risky_domains
```
