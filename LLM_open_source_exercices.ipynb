# 🌟 Exercice 1 : Réflexion sur les niveaux de l'open source

**Objectif :** définir les trois niveaux d'ouverture, les comparer, et répondre au cas santé.

````python
# Tableau comparatif des niveaux d'ouverture
niveaux = pd.DataFrame({
    "Niveau": ["Entièrement ouvert", "Poids publiés", "Architecture uniquement"],
    "Qu'est-ce qui est ouvert ?": [
        "Code + poids + données d'entraînement + architecture",
        "Poids pré-entraînés + architecture (ni données, ni pipeline complet)",
        "Structure / code du modèle uniquement (pas de poids)"
    ],
    "Impact sur modification / réentraînement": [
        "Inspection, modification et réentraînement complets possibles",
        "Fine-tuning possible ; reproduction impossible sans les données",
        "Réentraînement complet requis à partir de zéro"
    ]
})
display(niveaux)
````

### 📝 Paragraphe comparatif

Un modèle **entièrement ouvert** publie tout — code, architecture, poids et données d'entraînement — permettant un audit complet et une reproductibilité totale ; son avantage principal est la transparence, au prix d'exigences élevées en ressources pour tout reconstruire. Un modèle à **poids publiés** fournit les poids pré-entraînés et l'architecture, rendant le fine-tuning immédiatement possible ; sa limite est qu'on ne peut ni vérifier ni reproduire l'entraînement d'origine faute de données. Un modèle **architecture uniquement** révèle seulement la structure : utile pédagogiquement, mais inexploitable sans un entraînement complet coûteux. Le choix dépend donc de l'équilibre entre transparence recherchée et ressources disponibles.

### 🏥 Réponse — cas santé

Pour un assistant santé devant être réentraîné sur des données cliniques, le niveau **poids publiés** est au minimum essentiel (idéalement *entièrement ouvert*). Ce niveau fournit les poids pré-entraînés nécessaires au fine-tuning spécifique au domaine, évitant un entraînement de zéro tout en permettant l'adaptation aux données médicales sensibles.

# 🌟 Exercice 4 : Audit de préparation locale

**Objectif :** vérifier si votre machine peut exécuter un modèle 7B quantifié via `llama.cpp`.

La cellule ci-dessous **détecte automatiquement** les specs de la machine Colab (ou locale).

````python
import shutil, os, subprocess

def run(cmd):
    try:
        return subprocess.check_output(cmd, shell=True, text=True, stderr=subprocess.DEVNULL).strip()
    except Exception:
        return "N/A"

# --- RAM ---
try:
    ram_gb = round(os.sysconf("SC_PAGE_SIZE") * os.sysconf("SC_PHYS_PAGES") / (1024**3), 1)
except Exception:
    ram_gb = None

# --- Disque ---
total, used, free = shutil.disk_usage("/")
free_gb = round(free / (1024**3), 1)

# --- OS ---
os_info = run("uname -sr")

# --- CPU : jeux d'instructions ---
flags = run("grep -m1 -o -E 'avx512|avx2|avx|sse4_2' /proc/cpuinfo | sort -u | tr '\n' ' '") or "N/A"

# --- Compilateurs / build tools ---
gcc = run("gcc --version | head -1")
cmake = run("cmake --version | head -1")

print(f"RAM totale        : {ram_gb} GB")
print(f"Disque libre      : {free_gb} GB")
print(f"OS                : {os_info}")
print(f"Jeux d'instr. CPU : {flags}")
print(f"Compilateur       : {gcc}")
print(f"CMake             : {cmake}")
````

````python
# Tableau d'audit automatique
def check(ok):
    return "✅" if ok else "❌"

audit = pd.DataFrame({
    "Requirement": ["RAM (≥ 16 GB)", "Free Disk (≥ 40 GB)", "OS (Linux/WSL2)",
                    "AVX/AVX2 (llama.cpp)", "Compilateur C/C++"],
    "Your System Specs": [
        f"{ram_gb} GB", f"{free_gb} GB", os_info,
        flags.strip(), gcc if gcc != "N/A" else "absent"
    ],
    "Meets Requirement?": [
        check(ram_gb and ram_gb >= 16),
        check(free_gb >= 40),
        check("Linux" in os_info),
        check(any(x in flags for x in ["avx", "avx2", "avx512"])),
        check(gcc != "N/A"),
    ],
})
display(audit)

print("\n📦 Un modèle 7B quantifié Q4_K_M pèse ~4 Go et tourne sur ~6-8 Go de RAM.")
print("🔧 Mises à niveau selon les ❌ :")
print("   • RAM  : ajouter des modules (swap = lent, déconseillé)")
print("   • Disk : nettoyer / stockage externe")
print("   • OS   : installer Linux ou activer WSL2 (wsl --install)")
````
