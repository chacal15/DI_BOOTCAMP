 Exercice 1 : Création et manipulation de tableaux

```
import numpy as np

# Création du tableau de 0 à 9
tableau = np.arange(10)

print(tableau)
```

Exercice 2 : Conversion de type et opérations sur les tableaux


```
import numpy as np

# Conversion de la liste en tableau NumPy et changement de type
tableau_entier = np.array([3.14, 2.17, 0, 1, 2]).astype(int)

print(tableau_entier)
```

Exercice 3 : Travailler avec des tableaux multidimensionnels


```
import numpy as np

# Création du tableau 3x3 de 1 à 9
matrice_3x3 = np.arange(1, 10).reshape(3, 3)

print(matrice_3x3)
```

Exercice 4 : Création d'un tableau multidimensionnel avec des nombres aléatoires


```
import numpy as np

# Création du tableau 2D de forme (4, 5) avec des valeurs aléatoires
tableau_aleatoire = np.random.rand(4, 5)

print(tableau_aleatoire)
```

Exercice 5 : Indexation des tableaux

```
import numpy as np

# Tableau 2D d'exemple
tableau_2d = np.array([[10, 20, 30],
                       [40, 50, 60],
                       [70, 80, 90]])

# Sélection de la deuxième ligne (index 1)
deuxieme_ligne = tableau_2d[1]

print(deuxieme_ligne)
```

Exercice 6 : Inversion des éléments


```
import numpy as np

# Tableau 1D d'exemple
tableau = np.array([10, 20, 30, 40, 50])

# Inversion de l'ordre des éléments
tableau_inverse = tableau[::-1]

print(tableau_inverse)
```

Exercice 7 : Matrice identité

```
import numpy as np

# Création d'une matrice identité 4x4
matrice_identite = np.eye(4)

print(matrice_identite)
```

Exercice 8 : Fonctions agrégées simples

```
import numpy as np

# Tableau 1D d'exemple contenant les nombres de 0 à 9
tableau = np.arange(10)

# Calcul de la somme et de la moyenne
somme = tableau.sum()
moyenne = tableau.mean()

print(f"Sum: {somme}, Average: {moyenne}")
```

Exercice 9 : Créer un tableau et modifier sa structure

```
import numpy as np

# Création du tableau de 1 à 20 et remodelage en 4x5
matrice = np.arange(1, 21).reshape(4, 5)

print(repr(matrice))
```

Exercice 10 : Sélection conditionnelle des valeurs


```
import numpy as np

# Tableau d'exemple de 0 à 9
tableau = np.arange(10)

# Extraction des nombres impairs
impairs = tableau[tableau % 2 != 0]

print(repr(impairs))
```
