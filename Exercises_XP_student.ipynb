1.Quelle est la principale différence entre l'apprentissage automatique traditionnel et l'apprentissage profond ?*

L'apprentissage automatique traditionnel désigne un ensemble de méthodes statistiques qui apprennent à partir de caractéristiques  features  sélectionnées et préparées manuellement par un humain, tandis que l'apprentissage profond est une sous-catégorie de l'apprentissage automatique qui utilise des réseaux neuronaux à plusieurs couches capables d'apprendre automatiquement ces caractéristiques directement à partir des données brutes. Par exemple, pour classer des e-mails comme spam ou non-spam, une approche traditionnelle comme un arbre de décision nécessiterait qu'un ingénieur définisse manuellement des variables pertinentes présence de certains mots, nombre de liens, etc., alors qu'un réseau profond pourrait apprendre directement à partir du texte brut quelles combinaisons de mots ou de motifs sont révélatrices de spam.

2.Comment les réseaux neuronaux artificiels imitent-ils le cerveau humain ?**

Les réseaux neuronaux artificiels s'inspirent de la structure biologique du cerveau en organisant des unités de calcul appelées « neurones artificiels » en couches interconnectées, où chaque neurone reçoit des signaux, les pondère, puis transmet un résultat aux neurones suivants, de façon analogue à la manière dont les neurones biologiques communiquent via des synapses. Cette analogie reste cependant simplifiée : les neurones artificiels effectuent des opérations mathématiques sommes pondérées et fonctions d'activation bien plus rudimentaires que les processus électrochimiques complexes des neurones biologiques.

3.Pourquoi l'apprentissage profond est-il plus performant sur les grands ensembles de données que l'apprentissage automatique traditionnel ?

Les modèles d'apprentissage profond possèdent un grand nombre de paramètres ajustables, ce qui leur permet de capturer des relations complexes et non linéaires dans les données, mais cette capacité nécessite de grandes quantités d'exemples pour éviter le surapprentissage et converger vers des représentations utiles. Avec peu de données, les modèles traditionnels comme la régression logistique ou les forêts aléatoires ont tendance à mieux généraliser car ils comportent moins de paramètres, mais lorsque le volume de données augmente considérablement, les réseaux profonds peuvent exploiter cette richesse d'information pour apprendre des hiérarchies de caractéristiques de plus en plus abstraites, ce qui se traduit par de meilleures performances.

4.Quels sont les défis de l'apprentissage profond, et comment peut-on les relever ?

L'apprentissage profond présente plusieurs défis majeurs : un besoin important de données étiquetées, un coût de calcul élevé nécessitant souvent des GPU ou TPU, un risque de surapprentissage, un manque d'interprétabilité effet de « boîte noire , et une sensibilité aux biais présents dans les données d'entraînement. Pour y répondre, on peut utiliser l'augmentation de données et le transfert d'apprentissage pour pallier le manque de données, des techniques de régularisation dropout, arrêt précoce pour limiter le surapprentissage, des méthodes d'IA explicable comme SAHP ou les cartes de saillancepour améliorer l'interprétabilité, et une curation rigoureuse des jeux de données pour réduire les biais.

5.Qu'est-ce que l'ingénierie des caractéristiques, et pourquoi n'est-elle pas nécessaire en apprentissage profond ?**

L'ingénierie des caractéristiques est le processus par lequel un expert humain sélectionne, transforme ou crée manuellement des variables pertinentes à partir des données brutes afin d'améliorer la performance d'un modèle d'apprentissage automatique traditionnel. Elle n'est généralement pas nécessaire en apprentissage profond parce que les couches successives du réseau apprennent automatiquement à extraire et à combiner les caractéristiques pertinentes directement à partir des données brutes, ce qui élimine cette étape manuelle souvent longue et dépendante de l'expertise du domaine.

6.Quel rôle jouent les couches cachées dans un modèle d'apprentissage profond ?

Les couches cachées sont les couches intermédiaires situées entre la couche d'entrée et la couche de sortie d'un réseau neuronal, et elles ont pour rôle de transformer progressivement les données en représentations de plus en plus abstraites et utiles à la tâche visée. Par exemple, dans un réseau de reconnaissance d'images, les premières couches cachées peuvent apprendre à détecter des contours et des textures simples, les couches intermédiaires peuvent combiner ces éléments pour reconnaître des formes comme un œil ou une roue, et les couches plus profondes peuvent assembler ces formes pour identifier des objets complets un visage ou une voiture.

7.Dans un réseau neuronal artificiel, quel est le rôle d'une fonction d'activation ?

Une fonction d'activation est une fonction mathématique appliquée à la sortie d'un neurone qui introduit une non-linéarité dans le réseau, permettant ainsi au modèle d'apprendre des relations complexes plutôt que de se limiter à de simples combinaisons linéaires des entrées. Sans fonction d'activation non linéaire comme ReLU, sigmoïde ou tangente hyperbolique, un réseau neuronal, même avec de nombreuses couches, se comporterait essentiellement comme un simple modèle linéaire, incapable de résoudre des problèmes complexes comme la reconnaissance d'images ou la traduction automatique.

```python
# PREFILLED: just execute
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.datasets import mnist

print("TensorFlow:", tf.__version__)

# Load data
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Normalize to [0,1]
x_train = x_train.astype("float32") / 255.0
x_test  = x_test.astype("float32") / 255.0

# Flatten will be added in the model; keep images as 28x28 for now
print("Train shape:", x_train.shape, "Test shape:", x_test.shape)
```

```python
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.datasets import mnist # Import mnist

# Load data again to ensure y_train and y_test are defined, in case the previous cell wasn't run.
(x_train, y_train), (x_test, y_test) = mnist.load_data()

y_train_oh = to_categorical(y_train, 10)
y_test_oh  = to_categorical(y_test, 10)
print("Labels are: ", y_train_oh.shape, y_test_oh.shape)
```

```python
from tensorflow.keras import models, layers

model = models.Sequential([
    layers.Flatten(input_shape=(28, 28)),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax'),
])
```

```python
# To-Do: compile the model with adam, categorical_crossentropy, and accuracy
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
# To-Do: train then evaluate
history = model.fit(x_train, y_train_oh, epochs=10, validation_split=0.2)
test_loss, test_acc = model.evaluate(x_test, y_test_oh)
print({"test_loss": float(test_loss), "test_acc": float(test_acc)})
```

## 🌟 Exercise 5 (optional): Implementing Forward and Backward Propagation in Python

**As stated in the enoncé**  
You will code a simple neural network that performs forward propagation and backpropagation for a regression problem predicting exam scores from study hours.

**Guidance**  
Run the prefilled code. Then modify the learning rate or initial weights and observe changes in the loss and updates.
Provide a short written explanation.

**Learning point**  
Gradient descent updates parameters in the direction of steepest loss reduction.
The learning rate controls step size. Too small slows learning. Too large risks divergence.


```python
# PREFILLED: just execute
import numpy as np

# Initialize input data (features)
x = np.array([4, 80])  # 4 hours studied, previous test score: 80

# Initialize weights and bias
w = np.array([0.6, 0.3])  # Initial weights
b = 10  # Initial bias

# Forward Propagation
def forward_propagation(x, w, b):
    z = np.dot(x, w) + b  # Weighted sum
    return z  # Linear activation (No ReLU here, it's a regression task)

# Compute prediction
y_pred = forward_propagation(x, w, b)
y_true = 85  # Actual exam score

# Compute Loss (Mean Squared Error)
loss = 0.5 * (y_true - y_pred) ** 2

# Compute Gradients
grad_w = -(y_true - y_pred) * x  # Partial derivatives with respect to weights
grad_b = -(y_true - y_pred)  # Partial derivative with respect to bias

# Update Weights and Bias
learning_rate = 0.01
w_new = w - learning_rate * grad_w
b_new = b - learning_rate * grad_b

# Print Results
print("Initial Prediction:", y_pred)
print("Loss:", loss)
print("Updated Weights:", w_new)
print("Updated Bias:", b_new)
```

**To-Do:** Explain in full sentences why updating weights using gradient descent reduces the error. Then modify the learning rate or initial weights and describe how this affects learning.


**Interprétation**

La descente de gradient met à jour les poids et le biais dans la direction qui réduit l'erreur entre la prédiction et la valeur réelle. Le gradient indique la pente de la fonction de perte : en soustrayant ce gradient (multiplié par le taux d'apprentissage) du poids actuel, on déplace le poids dans la direction opposée à la pente, ce qui rapproche la prochaine prédiction de la valeur réelle et réduit donc la perte à l'itération suivante.

Le taux d'apprentissage contrôle la taille du pas effectué à chaque mise à jour. S'il est trop faible, les poids changent très peu à chaque itération, ce qui rend l'apprentissage extrêmement lent. S'il est trop élevé, les mises à jour deviennent trop brutales : le modèle peut dépasser le minimum de la fonction de perte, voire diverger, c'est-à-dire que la perte augmente au lieu de diminuer au fil des itérations. Un bon taux d'apprentissage doit donc être suffisamment grand pour progresser efficacement, mais suffisamment petit pour permettre une convergence stable.
